```python
def replace_words(roots, sentence):
    words = sentence.split(" ")
    #looping over each word 
    for index, word in enumerate(words):
        #looping over each root
        for root in roots:
            #checking if words start with roots
            if word.startswith(root):
                #replace word with its root
                words[index] = root
    return " ".join(words)           
```

```python
roots1 = ["cat", "bat", "rat"]
sentence1 = "the cattle was rattled by the battery"

replace_words(roots1, sentence1)
```
