```python
df1 = df[['customer_id', 'full_name', 'transaction_date', 'transaction_type', 'transaction_amount',
       'account_balance']]
```

```python
df1[df1['transaction_date']>"2024-01-01"]
```

```python
# filter out dataset for only transactions 
df1[df1["transaction_type"].isin(["ATM","PS"])]
```

```python
df1[df1['transaction_type'].isin(['ATM', 'PS'])]
```

```python
#ASSUME TRASACT_AMOUNT IS THE LAST TRANSACTION
# PREVIOUS TRANSACTION BALANCE IS transaction amount + account_balance
df1['previous_tran_balance'] = df1['transaction_amount'] + df1['account_balance']
df1
```

```python
#.agg method 

def sum_sta(column):
    return column.describe()
```

```python
df1['previous_tran_balance'].agg(sum_sta)
```
