```python
def suspic_count(df, allowed_list):
    allowed_list = set(allowed_list)
    sus_count_tran = 0
    #create the columns from the dictionary
    for record in df:
        amount = record["amount"]
        is_international = record["is_international"]
        country = record["country"]
        channel = record["channel"]

    #define conditions
        rule_1 = amount > 5000
        rule_2 =  is_international == True or amount > 1000
        rule_3 = channel == "online" and country not in allowed_list

    #check and count
        if rule_1 or rule_2 or rule_3:
            sus_count_tran += 1

        
    return  sus_count_tran






```

```python
import numpy as np

def dot_product(a, b):
    a_arr = np.array(a)
    b_arr = np.array(b)
    return float(np.dot(a_arr, b_arr))

# Test
print(dot_product([1, 2, 3], [4, 5, 6]))  # 32.0
```

```python
#17. Map Categorical Labels to Integers
#Task: Convert a list of string labels to integer IDs.
def encode_labels(labels):
    label_to_id = {}
    encoded = []
    next_id = 0
    for lab in labels:
        if lab not in label_to_id:
            label_to_id[lab] = next_id
            next_id += 1
        encoded.append(label_to_id[lab])
    return encoded, label_to_id

# Test
labels = ["spam", "ham", "spam", "egg"]
encoded, mapping = encode_labels(labels)
print(encoded)
print(mapping)
```
