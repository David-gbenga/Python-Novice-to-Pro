```python
#Create a Class with for a baby which is able to possess attributes like name, age,gender, citizenship, present_location.

class Baby_feat:
    
    def __init__ (self, name, age,gender,citizenship, present_location,tone):
        self.name = name
        self.age = age
        self.gender = gender
        self.citizenship = citizenship
        self.present_location = present_location
        self.tone = tone

    def cry_tone(self,tone):
        cry = tone *10
        return cry


#Also create a method for the baby where the baby can showcase their cry type
```

```python
baby_kay = Baby_feat("Ken",2,"Male","British","London","yenyen")
print(baby_kay.name)
```
